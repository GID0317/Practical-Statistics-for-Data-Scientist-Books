## Estimates of Location

Ukuran lokasi menjelaskan pusat data. Untuk $n$ pengamatan, mean adalah $\bar{x}=\sum x_i/n$, sedangkan median adalah nilai tengah setelah data diurutkan. Mean sensitif terhadap nilai ekstrem; median lebih tahan terhadap outlier. Trimmed mean membuang proporsi tertentu pada **masing-masing ekor**, kemudian merata-ratakan sisanya. `trim_mean(..., 0.1)` membuang 10% terbawah dan 10% teratas, bukan 10% total.

**Hubungan dengan kode:** populasi negara bagian digunakan untuk membandingkan ketiga ukuran. Jika mean lebih besar daripada median, beberapa negara bagian berpopulasi besar dapat menarik mean ke atas.

### Example: Location Estimates of Population and Murder Rates

Rata-rata sederhana memberi setiap negara bagian bobot sama. Weighted mean, $\bar{x}_w=\sum w_i x_i/\sum w_i$, memberi kontribusi sesuai bobot. Untuk `Murder.Rate`, bobot populasi mendekati tingkat pembunuhan gabungan per 100.000 penduduk, bukan rata-rata negara bagian. Weighted median mencari titik tengah massa bobot. Pilihan bobot harus mengikuti pertanyaan analisis, bukan dipilih agar hasil terlihat baik.

```python
# Table 1-2
state = pd.read_csv(STATE_CSV)
print(state.head(8))
```

## Estimates of Variability

Variabilitas menunjukkan seberapa tersebar data. Varians sampel $s^2=\sum(x_i-\bar{x})^2/(n-1)$; standar deviasi $s$ kembali ke satuan asal. IQR adalah $Q_{0.75}-Q_{0.25}$ dan berfokus pada 50% data tengah. MAD mentah adalah $\mathrm{median}(|x_i-\mathrm{median}(x)|)$.

`statsmodels.robust.scale.mad` secara default membagi MAD mentah dengan sekitar 0,67449 untuk konsistensi dengan standar deviasi pada distribusi normal. Karena itu hasilnya berbeda dari MAD mentah. SD sensitif terhadap outlier, sedangkan IQR dan MAD lebih robust.

```python
# Table 1-2
print(state.head(8))
```

```python
print(state['Population'].std())
```

```python
print(state['Population'].quantile(0.75) - state['Population'].quantile(0.25))
```

```python
print(robust.scale.mad(state['Population']))
print(abs(state['Population'] - state['Population'].median()).median() / 0.6744897501960817)
```

```python
from matplotlib.collections import EllipseCollection
from matplotlib.colors import Normalize

def plot_corr_ellipses(data, figsize=None, **kwargs):
    ''' https://stackoverflow.com/a/34558488 '''
    M = np.array(data)
    if not M.ndim == 2:
        raise ValueError('data must be a 2D array')
    fig, ax = plt.subplots(1, 1, figsize=figsize, subplot_kw={'aspect':'equal'})
    ax.set_xlim(-0.5, M.shape[1] - 0.5)
    ax.set_ylim(-0.5, M.shape[0] - 0.5)
    ax.invert_yaxis()

    # xy locations of each ellipse center
    xy = np.indices(M.shape)[::-1].reshape(2, -1).T

    # set the relative sizes of the major/minor axes according to the strength of
    # the positive/negative correlation
    w = np.ones_like(M).ravel() + 0.01
    h = 1 - np.abs(M).ravel() - 0.01
    a = 45 * np.sign(M).ravel()

    ec = EllipseCollection(widths=w, heights=h, angles=a, units='x', offsets=xy,
                           norm=Normalize(vmin=-1, vmax=1),
                           transOffset=ax.transData, array=M.ravel(), **kwargs)
    ax.add_collection(ec)

    # if data is a DataFrame, use the row/column names as tick labels
    if isinstance(data, pd.DataFrame):
        ax.set_xticks(np.arange(M.shape[1]))
        ax.set_xticklabels(data.columns, rotation=90)
        ax.set_yticks(np.arange(M.shape[0]))
        ax.set_yticklabels(data.index)

    return ec, ax

m, ax = plot_corr_ellipses(etfs.corr(), figsize=(5, 4), cmap='bwr_r')
cb = plt.colorbar(m, ax=ax)
cb.set_label('Correlation coefficient')

plt.tight_layout()
plt.show()
```

### Two Categorical Variables

```python
lc_loans = pd.read_csv(LC_LOANS_CSV)
```

## Rectangular Data, Data Frames, dan Indexes

Data rectangular berbentuk tabel: satu baris mewakili satu record/pengamatan, satu kolom mewakili satu variabel. Dalam pembelajaran mesin, prediktor juga disebut **feature** dan respons disebut **target**. Istilah sample dalam statistik berarti sekumpulan pengamatan, bukan selalu satu baris.

`pandas.DataFrame` menyimpan tabel beserta label kolom dan index baris. Index membantu memilih serta menyelaraskan pengamatan; tidak otomatis merupakan feature untuk model. Dua tabel harus digabungkan berdasarkan kunci yang benar, bukan sekadar posisi baris. ID pelanggan atau kode pos tidak menjadi variabel kontinu hanya karena tersimpan sebagai angka. Pada matriks sparse, sebagian besar elemen nol; penyimpanan sparse dapat lebih hemat daripada tabel padat.

### Nonrectangular Data Structures

- **Time series:** pengamatan berurutan menurut waktu. Urutan dan ketergantungan waktu harus dipertahankan saat analisis atau validasi.
- **Spatial data:** merepresentasikan objek dengan koordinat atau nilai pada sel wilayah/raster. Objek berdekatan dapat mempunyai pengamatan yang berkaitan.
- **Graph/network:** node mewakili entitas dan edge mewakili hubungan, misalnya pengguna dan pertemanan. Grafik visualisasi distribusi tidak sama dengan struktur data graph.

Data tersebut dapat menghasilkan fitur tabular, tetapi proses konversi tidak boleh menghapus struktur penting. Bab ini terutama memakai data rectangular; tidak memerlukan implementasi forecasting, GIS, atau graph model.

## Mode, Expected Value, dan Probability

### Mode

Mode adalah nilai dengan frekuensi tertinggi. Jika beberapa nilai mempunyai frekuensi maksimum sama, data mempunyai lebih dari satu mode. Mean tidak bermakna untuk kategori nominal seperti penyebab keterlambatan; mode dapat merangkumnya. Untuk data kontinu, banyak nilai unik membuat mode mentah kurang berguna dibanding histogram/KDE. Mode pada data yang dibulatkan juga dipengaruhi ketelitian pengukuran.

### Expected Value

Untuk hasil diskrit $x_i$ dengan probabilitas $p_i$, expected value adalah $E[X]=\sum_i p_i x_i$, dengan $p_i\geq0$ dan $\sum_i p_i=1$. Ini rata-rata jangka panjang di bawah model probabilitas, bukan hasil yang pasti didapat satu individu. Contoh tambahan: peluang pendapatan 100, 40, dan 0 masing-masing 0,10; 0,20; dan 0,70 memberikan expected value 18 per pelanggan. Expected value saja belum menjelaskan risiko atau variasi outcome.

### Probability

Probabilitas bernilai antara nol dan satu. Frekuensi relatif dari banyak pengulangan dapat mengestimasi probabilitas jika proses dan sampelnya sesuai. Untuk kejadian komplementer, $P(A^c)=1-P(A)$. Probabilitas bersyarat $P(A\mid B)=P(A\cap B)/P(B)$ membutuhkan denominator kelompok B. Pada pinjaman, peluang charged-off di antara grade A berbeda dari proporsi seluruh pinjaman yang sekaligus grade A dan charged-off. Odds sukses adalah $p/(1-p)$; odds a:b setara probabilitas $a/(a+b)$.

```python
# Ilustrasi tambahan; bukan dataset atau angka yang disalin dari buku.
mode_example = pd.Series(['A', 'B', 'B', 'C', 'C'])
print('Semua mode:', mode_example.mode().tolist())
ev_outcomes = np.array([100.0, 40.0, 0.0])
ev_probabilities = np.array([0.10, 0.20, 0.70])
expected_revenue = np.sum(ev_outcomes * ev_probabilities)
print('Expected revenue per pelanggan:', expected_revenue)
assert mode_example.mode().tolist() == ['B', 'C']
assert np.isclose(ev_probabilities.sum(), 1)
assert np.isclose(expected_revenue, 18)
print('Interpretasi: rata-rata model 18; bukan pendapatan pasti tiap pelanggan.')
```

## Pelengkap Ukuran Variabilitas

**Range** adalah $\max(x)-\min(x)$ dan bergantung hanya pada dua nilai ekstrem. **Mean absolute deviation** adalah $\sum|x_i-\bar{x}|/n$, berbeda dari median absolute deviation dari median. Varians/SD memberi penalti kuadrat sehingga nilai ekstrem lebih berpengaruh daripada pada deviasi absolut. **Order statistics** adalah nilai setelah diurutkan; median, kuartil, dan persentil berasal dari urutan tersebut. Interpolasi kuantil pada sampel terbatas dapat berbeda antar-library.

Pemilihan mean/SD atau median/IQR bergantung pada bentuk distribusi dan tujuan. Robust berarti relatif tahan terhadap kontaminasi nilai ekstrem, bukan kebal terhadap seluruh masalah data. Nilai missing tidak sama dengan nol; keputusan membuang atau mengimputasinya perlu mempertimbangkan alasan nilai hilang. Sel sumber memakai estimasi pada data tersedia, bukan pembenaran otomatis untuk imputasi.

## Ringkasan bab

Ukuran pusat dan sebaran perlu dibaca bersama. Pada data populasi negara bagian, mean sekitar **6,16 juta** lebih besar daripada median **4,44 juta** karena negara bagian berpopulasi besar menarik rata-rata ke atas. Trimmed mean sekitar **4,78 juta** lebih dekat ke median setelah kedua ekor data dipangkas. Perbandingan ini menunjukkan manfaat ukuran yang lebih tahan terhadap nilai ekstrem.

Persentase pinjaman berstatus **Charged Off** per grade memakai jumlah pinjaman pada grade tersebut sebagai penyebut. Hasilnya sekitar **2,15% untuk grade A** dan **12,62% untuk grade G**. Ini berbeda dari persentase terhadap seluruh pinjaman, yang juga dipengaruhi banyaknya pinjaman di setiap grade. Pada data rumah, pemisahan grafik menurut kode pos membantu melihat pola lokasi yang dapat tersembunyi ketika semua rumah digabung. Kode pos merupakan label kategori; selisih angkanya tidak menyatakan jarak atau urutan kualitas lokasi.

Pada ilustrasi expected value, hasil **18** adalah rata-rata tertimbang probabilitas dari outcome 100, 40, dan 0. Nilai tersebut bukan hasil yang diterima setiap pelanggan. Mode menjawab kategori mana yang paling sering muncul, sedangkan expected value menjawab rata-rata outcome menurut probabilitasnya.

Pemilihan tabel dan grafik mengikuti jenis variabel serta kepadatan data. Median dan IQR membantu ketika nilai ekstrem kuat, pembobotan mengubah kontribusi setiap pengamatan, dan index membantu menghubungkan baris data dengan benar. Korelasi merangkum asosiasi linear; kesimpulan sebab-akibat memerlukan bukti tambahan.
