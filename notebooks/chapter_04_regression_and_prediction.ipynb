## Prediksi, inferensi, dan validasi

Prediksi berfokus pada akurasi untuk observasi baru; inferensi mempelajari hubungan parameter beserta ketidakpastiannya. Keduanya memerlukan pertanyaan yang jelas. ## Simple Linear Regression

Regresi memodelkan nilai numerik respons berdasarkan prediktor. Regresi linear sederhana memakai $Y=\beta_0+\beta_1X+\epsilon$. Ordinary least squares memilih koefisien yang meminimalkan jumlah residual kuadrat. Pada contoh, Exposure memprediksi PEFR; asosiasi tersebut sendiri tidak membuktikan sebab-akibat.

### The Regression Equation

Intercept adalah prediksi saat X nol; slope adalah perubahan prediksi Y untuk kenaikan X satu unit. Intercept dapat tidak bermakna praktis jika nol di luar rentang data. Tanda negatif koefisien Exposure menunjukkan prediksi PEFR menurun ketika Exposure meningkat. Hindari ekstrapolasi jauh dari rentang pengamatan.

Untuk X yang tidak konstan, solusi OLS sederhana dapat ditulis sebagai $b_1=\frac{\sum_i(x_i-\bar{x})(y_i-\bar{y})}{\sum_i(x_i-\bar{x})^2}$ dan $b_0=\bar{y}-b_1\bar{x}$. Penyebut mengukur variasi X; pembilang mengukur variasi bersama X dan Y. `LinearRegression.fit` menghitung solusi least squares secara numerik. Rumus ini menjelaskan asal slope, bukan hanya cara memanggil fungsi library.

```python
lung = pd.read_csv(LUNG_CSV)

lung.plot.scatter(x='Exposure', y='PEFR')

plt.tight_layout()
plt.show()
```

```python
predictors = ['Exposure']
outcome = 'PEFR'

model = LinearRegression()
model.fit(lung[predictors], lung[outcome])

print(f'Intercept: {model.intercept_:.3f}')
print(f'Coefficient Exposure: {model.coef_[0]:.3f}')
```

```python
fig, ax = plt.subplots(figsize=(4, 4))
ax.set_xlim(0, 23)
ax.set_ylim(295, 450)
ax.set_xlabel('Exposure')
ax.set_ylabel('PEFR')
ax.plot((0, 23), model.predict(pd.DataFrame({'Exposure': [0, 23]})))
ax.text(0.4, model.intercept_, r'$b_0$', size='larger')

x = pd.DataFrame({'Exposure': [7.5,17.5]})
y = model.predict(x)
ax.plot((7.5, 7.5, 17.5), (y[0], y[1], y[1]), '--')
ax.text(5, np.mean(y), r'$\Delta Y$', size='larger')
ax.text(12, y[1] - 10, r'$\Delta X$', size='larger')
ax.text(12, 390, r'$b_1 = \frac{\Delta Y}{\Delta X}$', size='larger')

plt.tight_layout()
plt.show()
```

### Fitted Values and Residuals

Fitted value $\hat{y}_i$ adalah prediksi model dan residual $e_i=y_i-\hat{y}_i$ adalah selisih observasi terhadap prediksi. Garis vertikal pada grafik memperlihatkan residual. Residual positif berarti model memprediksi terlalu rendah. Residual mengestimasi error, tetapi tidak identik dengan error populasi yang tidak teramati.

```python
fitted = model.predict(lung[predictors])
residuals = lung[outcome] - fitted
```

```python
ax = lung.plot.scatter(x='Exposure', y='PEFR', figsize=(4, 4))
ax.plot(lung.Exposure, fitted)
for x, yactual, yfitted in zip(lung.Exposure, lung.PEFR, fitted): 
    ax.plot((x, x), (yactual, yfitted), '--', color='C1')

plt.tight_layout()
plt.show()
```

## Multiple Linear Regression

Regresi berganda memakai $\hat{Y}=b_0+\sum_j b_jX_j$. Koefisien satu prediktor ditafsirkan dengan prediktor lain tetap. Harga rumah diprediksi dari luas bangunan/tanah, kamar mandi, kamar tidur, dan grade. Satuan tiap koefisien mengikuti satuan prediktornya, sehingga besar koefisien mentah tidak langsung mengukur kepentingan relatif.

```python
subset = ['AdjSalePrice', 'SqFtTotLiving', 'SqFtLot', 'Bathrooms', 
          'Bedrooms', 'BldgGrade']

house = pd.read_csv(HOUSE_CSV, sep='\t')
print(house[subset].head())
```

```python
predictors = ['SqFtTotLiving', 'SqFtLot', 'Bathrooms', 
              'Bedrooms', 'BldgGrade']
outcome = 'AdjSalePrice'

house_lm = LinearRegression()
house_lm.fit(house[predictors], house[outcome])

print(f'Intercept: {house_lm.intercept_:.3f}')
print('Coefficients:')
for name, coef in zip(predictors, house_lm.coef_):
    print(f' {name}: {coef}')
```

### Assessing the Model

RMSE adalah $\sqrt{\sum(y_i-\hat{y}_i)^2/n}$ dalam satuan harga. $R^2=1-SSE/SST$ mengukur perbaikan terhadap prediksi mean. Metrik pada contoh sumber dihitung pada data training, sehingga tidak mengukur kemampuan generalisasi. Ringkasan OLS memuat standard error, t-statistic, interval, dan p-value yang bergantung pada asumsi model. `statsmodels` memerlukan intercept eksplisit.

**RMSE versus residual standard error (RSE):** keduanya mengukur ukuran residual dalam satuan respons, tetapi denominator berbeda. Dengan n pengamatan, p prediktor dan satu intercept pada OLS full rank, $RMSE=\sqrt{RSS/n}$ sedangkan $RSE=\sqrt{RSS/(n-p-1)}$. RSE memperhitungkan derajat bebas yang digunakan mengestimasi koefisien; keduanya makin dekat jika n jauh lebih besar daripada p. RSE training bukan estimasi otomatis error pada data test.

**Adjusted R-squared:** $R^2_{adj}=1-(1-R^2)\frac{n-1}{n-p-1}$ untuk model dengan intercept. Menambah prediktor pada OLS nested di data training yang sama tidak menurunkan R² biasa, tetapi adjusted R² dapat turun karena penalti kompleksitas. Metrik ini tetap in-sample dan tidak menggantikan cross-validation/holdout. R² pada test dapat negatif ketika model lebih buruk daripada baseline mean test yang dipakai definisi R².

**Standard error koefisien:** SE merangkum ketidakpastian estimasi b_j. Untuk hipotesis koefisien nol, statistik t=b_j/SE(b_j) menghasilkan p-value berdasarkan model error dan derajat bebas. SE koefisien berbeda dari SD respons, RSE residual, maupun prediction interval untuk satu observasi baru.

```python
fitted = house_lm.predict(house[predictors])
RMSE = np.sqrt(mean_squared_error(house[outcome], fitted))
r2 = r2_score(house[outcome], fitted)
print(f'RMSE: {RMSE:.0f}')
print(f'r2: {r2:.4f}')
```

```python
model = sm.OLS(house[outcome], house[predictors].assign(const=1))
results = model.fit()
print(results.summary())
```

### Model Selection and Stepwise Regression

AIC menyeimbangkan likelihood dan kompleksitas: $AIC=2k-2\log L$; nilai lebih kecil lebih disukai untuk model pada respons dan data yang sama. Stepwise menambah/menghapus prediktor secara bertahap menurut skor, bukan mengevaluasi seluruh kemungkinan model. Pilihan bisa tidak stabil pada prediktor berkorelasi. P-value setelah seleksi tidak memiliki interpretasi sederhana seolah model sudah ditetapkan sebelum melihat data.

```python
predictors = ['SqFtTotLiving', 'SqFtLot', 'Bathrooms', 'Bedrooms',
              'BldgGrade', 'PropertyType', 'NbrLivingUnits',
              'SqFtFinBasement', 'YrBuilt', 'YrRenovated', 
              'NewConstruction']

X = pd.get_dummies(house[predictors], drop_first=True, dtype=int)
X['NewConstruction'] = [1 if nc else 0 for nc in X['NewConstruction']]

house_full = sm.OLS(house[outcome], X.assign(const=1))
results = house_full.fit()
print(results.summary())
```

```python
y = house[outcome]

def train_model(variables):
    if len(variables) == 0:
        return None
    model = LinearRegression()
    model.fit(X[variables], y)
    return model

def score_model(model, variables):
    if len(variables) == 0:
        return AIC_score(y, [y.mean()] * len(y), model, df=1)
    return AIC_score(y, model.predict(X[variables]), model)

best_model, best_variables = stepwise_selection(X.columns, train_model, score_model, 
                                                verbose=True)

print()
print(f'Intercept: {best_model.intercept_:.3f}')
print('Coefficients:')
for name, coef in zip(best_variables, best_model.coef_):
    print(f' {name}: {coef}')
```

### Weighted Regression

Weighted least squares meminimalkan $\sum w_i(y_i-\hat{y}_i)^2$. Contoh memberi transaksi lebih baru bobot lebih besar dengan `Year - 2005`. Ini adalah pilihan desain untuk menekankan data baru; bobot tersebut tidak otomatis inverse variance. Pembobotan membuat hasil fit lebih dipengaruhi transaksi baru, sehingga koefisien dapat berbeda dari regresi tanpa bobot. Pilihan ini sesuai tujuan memberi relevansi lebih besar pada transaksi baru, tetapi tidak dengan sendirinya mengatasi heteroskedastisitas atau menjamin prediksi lebih akurat.

```python
house['Year'] = [int(date.split('-')[0]) for date in house.DocumentDate]
house['Year'] = house.DocumentDate.apply(lambda d: int(d.split('-')[0]))
house['Weight'] = house.Year - 2005
```

```python
predictors = ['SqFtTotLiving', 'SqFtLot', 'Bathrooms', 
              'Bedrooms', 'BldgGrade']
outcome = 'AdjSalePrice'

house_wt = LinearRegression()
house_wt.fit(house[predictors], house[outcome], sample_weight=house.Weight)

pd.concat([
    pd.DataFrame({
        'predictor': predictors,
        'house_lm': house_lm.coef_,
        'house_wt': house_wt.coef_,    
    }),
    pd.DataFrame({
        'predictor': ['intercept'],
        'house_lm': house_lm.intercept_,
        'house_wt': house_wt.intercept_,    
    })
])
```

```python
residuals = pd.DataFrame({
    'abs_residual_lm': np.abs(house_lm.predict(house[predictors]) - house[outcome]),
    'abs_residual_wt': np.abs(house_wt.predict(house[predictors]) - house[outcome]),
    'Year': house['Year'],
})
print(residuals.head())
# axes = residuals.boxplot(['abs_residual_lm', 'abs_residual_wt'], by='Year', figsize=(10, 4))
# axes[0].set_ylim(0, 300000)

pd.DataFrame(([year, np.mean(group['abs_residual_lm']), np.mean(group['abs_residual_wt'])] 
              for year, group in residuals.groupby('Year')),
             columns=['Year', 'mean abs_residual_lm', 'mean abs_residual_wt'])
# for year, group in residuals.groupby('Year'):
#     print(year, np.mean(group['abs_residual_lm']), np.mean(group['abs_residual_wt']))
```

### Factor Variables with Many Levels

Kode pos mempunyai banyak level. Contoh mengelompokkan kode pos berdasarkan median residual dan jumlah kumulatif rumah. Karena pengelompokan memakai target harga, membentuknya sebelum train/test split menyebabkan **target leakage**. Bagian ini direproduksi untuk pembelajaran pada data training; evaluasi holdout tambahan di akhir menggunakan fitur numerik dan tidak memakai ZipGroup. Dalam pipeline produksi, pemetaan harus dipelajari hanya dari training.

```python
print(pd.DataFrame(house['ZipCode'].value_counts()).transpose())
```

```python
house = pd.read_csv(HOUSE_CSV, sep='\t')

predictors = ['SqFtTotLiving', 'SqFtLot', 'Bathrooms', 
              'Bedrooms', 'BldgGrade']
outcome = 'AdjSalePrice'

house_lm = LinearRegression()
house_lm.fit(house[predictors], house[outcome])


zip_residuals = pd.DataFrame({
    'ZipCode': house['ZipCode'],
    'residual' : house[outcome] - house_lm.predict(house[predictors]),
})

zip_groups = pd.DataFrame([
    {
        'ZipCode': zipCode,
        'count': len(x),
        'median_residual': x.residual.median()
    } 
    for zipCode, x in zip_residuals.groupby('ZipCode')
]).sort_values('median_residual')

zip_groups['cum_count'] = np.cumsum(zip_groups['count'])
zip_groups['ZipGroup'] = pd.qcut(zip_groups['cum_count'], 5, labels=False, retbins=False)
zip_groups.head()
print(zip_groups.ZipGroup.value_counts().sort_index())
```

```python
to_join = zip_groups[['ZipCode', 'ZipGroup']].set_index('ZipCode')
house = house.join(to_join, on='ZipCode')
house['ZipGroup'] = house['ZipGroup'].astype('category')
```

Confidence interval parameter berbeda dari prediction interval untuk satu rumah baru; interval prediksi memperhitungkan variasi individual dan biasanya lebih lebar.

### Confounding Variables

Variabel yang berkaitan dengan prediktor dan outcome dapat mengubah estimasi asosiasi ketika tidak dimasukkan. Penambahan lokasi pada harga rumah dapat mengubah koefisien luas. Namun memasukkan beberapa kovariat tidak otomatis menjamin semua confounding teratasi; klaim kausal memerlukan desain dan asumsi tambahan.

```python
predictors = ['SqFtTotLiving', 'SqFtLot', 'Bathrooms', 'Bedrooms',
              'BldgGrade', 'PropertyType', 'ZipGroup']
outcome = 'AdjSalePrice'

X = pd.get_dummies(house[predictors], drop_first=True)

confounding_lm = LinearRegression()
confounding_lm.fit(X, house[outcome])

print(f'Intercept: {confounding_lm.intercept_:.3f}')
print('Coefficients:')
for name, coef in zip(X.columns, confounding_lm.coef_):
    print(f' {name}: {coef}')
```

### Interactions and Main Effects

Interaksi berarti efek satu prediktor bergantung pada prediktor lain. Formula `SqFtTotLiving*ZipGroup` mencakup main effects dan interaksi. Slope luas rumah dapat berbeda menurut kelompok lokasi. Main effect luas ditafsirkan pada kategori acuan ZipGroup; jangan membacanya sebagai slope yang berlaku sama di semua lokasi.

```python
model = smf.ols(formula='AdjSalePrice ~  SqFtTotLiving*ZipGroup + SqFtLot + ' +
     'Bathrooms + Bedrooms + BldgGrade + PropertyType', data=house)
results = model.fit()
print(results.summary())
```

## Testing the Assumptions: Regression Diagnostics

Diagnostik mengevaluasi linearitas bentuk mean, independensi error, variasi error, dan distribusi residual. Normalitas error terutama relevan untuk inferensi tertentu, bukan syarat bahwa seluruh prediktor atau outcome harus normal. Grafik residual dapat memandu revisi model; satu metrik training tidak cukup.

### Outliers

Outlier respons memiliki residual besar. Studentized residual menskalakan residual berdasarkan variasi dan leverage. Contoh memakai residual studentized internal. Data ekstrem perlu diperiksa asalnya; jangan otomatis dihapus agar model tampak lebih baik.

### Influential values

Leverage menilai posisi ekstrem pada ruang prediktor. Cook's distance menilai perubahan model akibat penghilangan sebuah pengamatan. Outlier, leverage tinggi, dan influential point saling berkaitan tetapi tidak sama. Contoh removal dengan cutoff 0,08 adalah analisis sensitivitas untuk membandingkan koefisien, bukan aturan universal pembersihan data.

```python
%matplotlib inline
from scipy.stats import linregress

np.random.seed(5)
x = np.random.normal(size=25)
y = -x / 5 + np.random.normal(size=25)
x[0] = 8
y[0] = 8

def abline(slope, intercept, ax):
    """Calculate coordinates of a line based on slope and intercept"""
    x_vals = np.array(ax.get_xlim())
    return (x_vals, intercept + slope * x_vals)

fig, ax = plt.subplots(figsize=(4, 4))
ax.scatter(x, y)
slope, intercept, _, _, _ = linregress(x, y)
ax.plot(*abline(slope, intercept, ax))
slope, intercept, _, _, _ = linregress(x[1:], y[1:])
ax.plot(*abline(slope, intercept, ax), '--')
ax.set_xlim(-2.5, 8.5)
ax.set_ylim(-2.5, 8.5)

plt.tight_layout()
plt.show()
```

```python
influence = OLSInfluence(result_98105)
fig, ax = plt.subplots(figsize=(5, 5))
ax.axhline(-2.5, linestyle='--', color='C1')
ax.axhline(2.5, linestyle='--', color='C1')
ax.scatter(influence.hat_matrix_diag, influence.resid_studentized_internal, 
           s=1000 * np.sqrt(influence.cooks_distance[0]),
           alpha=0.5)

ax.set_xlabel('hat values')
ax.set_ylabel('studentized residuals')

plt.tight_layout()
plt.show()
```

```python
mask = [dist < .08 for dist in influence.cooks_distance[0]]
house_infl = house_98105.loc[mask]

ols_infl = sm.OLS(house_infl[outcome], house_infl[predictors].assign(const=1))
result_infl = ols_infl.fit()

pd.DataFrame({
    'Original': result_98105.params,
    'Influential removed': result_infl.params,
})
```

### Heteroskedasticity, Non-Normality and Correlated Errors

Heteroskedastisitas berarti varians error tidak konstan. Pola naik pada absolute residual versus fitted value dapat menjadi indikasi. OLS standard error klasik dapat tidak sesuai; robust standard error atau pemodelan varians dapat dipertimbangkan sesuai tujuan. Error berkorelasi umum pada data waktu/kelompok. Histogram residual hanya memberi petunjuk bentuk distribusi, bukan uji independensi.

### Polynomial and Spline Regression

Regresi polinomial menambahkan pangkat prediktor, misalnya $X^2$. Model tetap linear terhadap koefisien meskipun nonlinear terhadap X. Fleksibilitas meningkat, tetapi derajat tinggi dapat overfit dan berperilaku buruk saat ekstrapolasi. Koefisien X harus ditafsirkan bersama koefisien pangkatnya.

```python
model_poly = smf.ols(formula='AdjSalePrice ~  SqFtTotLiving + np.power(SqFtTotLiving, 2) + ' + 
                'SqFtLot + Bathrooms + Bedrooms + BldgGrade', data=house_98105)
result_poly = model_poly.fit()
print(result_poly.summary())
```

```python
def partialResidualPlot(model, df, outcome, feature, ax):
    y_pred = model.predict(df)
    # determine columns required for model
    required = set(model.params.index).intersection(df.columns)
    required.add(feature)
    copy_df = df[list(required)].copy().astype('float')
    for c in copy_df.columns:
        if c == feature:
            continue
        copy_df.loc[:, c] = 0.0
    feature_prediction = model.predict(copy_df)
    results = pd.DataFrame({
        'feature': df[feature],
        'residual': df[outcome] - y_pred,
        'ypartial': feature_prediction - model.params.iloc[0],
    })
    results = results.sort_values(by=['feature'])
    smoothed = sm.nonparametric.lowess(results.ypartial + results.residual,
                                       results.feature, frac=1/3)
    
    ax.scatter(results.feature, results.ypartial + results.residual)
    ax.plot(smoothed[:, 0], smoothed[:, 1], color='gray')
    ax.plot(results.feature, results.ypartial, color='black')
    ax.set_xlabel(feature)
    ax.set_ylabel(f'Residual + {feature} contribution')
    return ax

fig, ax = plt.subplots(figsize=(5, 5))
partialResidualPlot(result_poly, house_98105, 'AdjSalePrice', 'SqFtTotLiving', ax)

plt.tight_layout()
plt.show()
print(result_poly.params.iloc[2])
```

### Generalized Additive Models

GAM mengganti sejumlah kontribusi linear dengan fungsi smooth: $E[Y|X]=\beta_0+\sum_j f_j(X_j)$ untuk link identitas. Model tetap additive kecuali interaksi ditambahkan. Penalti mengendalikan kekasaran fungsi. Output kedua library tidak harus identik karena basis, penalti, dan spesifikasi berbeda.

#### Using statsmodels

GLMGam menggunakan basis spline untuk luas bangunan, sedangkan luas tanah, kamar mandi, kamar tidur, dan grade tetap linear. Kode sumber memasukkan komponen linear dan spline pada fitur yang sama sehingga matriks desain rank-deficient: beberapa koefisien tidak teridentifikasi secara unik. Adaptasi ini menghilangkan redundansi, lalu memeriksa rank matriks desain.

Basis spline tidak menyertakan intercept karena intercept sudah ada pada formula. Penalti alpha nol mengikuti demonstrasi sumber dan tidak menjamin kompleksitas optimal. Perbedaan hasil dengan pyGAM tetap mungkin karena basis dan pemilihan penalti berbeda.

```python
from statsmodels.gam.api import GLMGam, BSplines

# Spline pada luas bangunan; fitur lainnya linear. Jangan masukkan
# luas secara linear lagi karena akan redundan dengan basis spline.
x_spline = house_98105[['SqFtTotLiving']]
bs = BSplines(x_spline, df=[10], degree=[3], include_intercept=False)
alpha = np.array([0.0])  # mengikuti demonstrasi tanpa penalti dari sumber
formula = 'AdjSalePrice ~ SqFtLot + Bathrooms + Bedrooms + BldgGrade'
gam_sm = GLMGam.from_formula(formula, data=house_98105, smoother=bs, alpha=alpha)
res_sm = gam_sm.fit()
print(res_sm.summary())
assert np.linalg.matrix_rank(gam_sm.exog) == gam_sm.exog.shape[1]
```

## Ordered Factor Variables

Kategori ordinal mempunyai urutan, misalnya rendah < sedang < tinggi. Urutan tidak otomatis menjamin jarak yang sama: beda rendah ke sedang belum tentu sama dengan sedang ke tinggi. `BldgGrade` pada contoh rumah merupakan kategori berurutan yang dipakai secara numerik pada model sumber.

Encoding angka mempertahankan urutan secara ringkas; dummy tidak memberlakukan urutan/jarak, meskipun efek hasil fit masih dapat diperiksa menurut urutan.

## Prediction Using Regression: Confidence dan Prediction Intervals

Prediksi titik pada fitur baru $x_0$ adalah $\hat{y}_0=x_0^T\hat\beta$. **Confidence interval koefisien** mengukur ketidakpastian parameter; **confidence interval mean response** mengukur ketidakpastian $E[Y\mid x_0]$; **prediction interval** mengukur ketidakpastian satu outcome baru $Y_{new}\mid x_0$. Ketiganya menjawab pertanyaan berbeda.

Untuk OLS klasik full rank dengan error independen normal dan homoskedastik, $SE_{mean}=s\sqrt{x_0^T(X^TX)^{-1}x_0}$ dan $SE_{pred}=s\sqrt{1+x_0^T(X^TX)^{-1}x_0}$. Interval menggunakan nilai kritis t dengan df residual. Tambahan 1 menggambarkan variasi individual yang masih ada walaupun koefisien diketahui tepat, sehingga prediction interval lebih lebar untuk titik dan tingkat cakupan sama.

Contoh tambahan memakai hubungan Exposure-PEFR. `mean_ci_*` pada statsmodels membatasi rata-rata respons; `obs_ci_*` membatasi satu observasi baru. Jangan menyebut CI mean sebagai rentang untuk 95% individu. Validitas interval tetap bergantung pada model; angka di bawah bukan penilaian medis.

### Bootstrap untuk Ketidakpastian Regresi

Untuk CI koefisien, resample pasangan baris fitur-target bersama-sama, refit model, lalu ambil kuantil koefisien. Untuk interval mean pada x0, simpan prediksi setiap refit. Untuk prediction interval, ketidakpastian individu juga perlu dimodelkan, misalnya menambah residual yang di-resample pada prediksi. Resampling residual biasa mengasumsikan struktur error sesuai; heteroskedastisitas, kelompok, dan waktu memerlukan prosedur yang lebih cocok.

### The Dangers of Extrapolation

Model yang dilatih pada rumah dengan bangunan tidak otomatis dapat menilai tanah kosong dengan semua fitur bangunan nol. Rentang setiap kolom secara terpisah juga belum memastikan kombinasi fitur pernah didukung data. Jauh dari dukungan training, prediksi dan interval berbasis model dapat menyesatkan. Confidence level tidak menyelesaikan kesalahan spesifikasi atau perubahan distribusi.

```python
interval_model = smf.ols('PEFR ~ Exposure', data=lung).fit()
interval_input = pd.DataFrame({'Exposure': [lung['Exposure'].median()]})
interval_output = interval_model.get_prediction(interval_input).summary_frame(alpha=0.05)
display(interval_input)
display(interval_output[['mean', 'mean_ci_lower', 'mean_ci_upper',
                         'obs_ci_lower', 'obs_ci_upper']])
mean_interval_width = interval_output['mean_ci_upper'] - interval_output['mean_ci_lower']
prediction_interval_width = interval_output['obs_ci_upper'] - interval_output['obs_ci_lower']
assert (prediction_interval_width > mean_interval_width).all()
print('Interpretasi: interval satu observasi baru lebih lebar daripada CI rata-rata respons.')
```

## Cross-Validation

Pada k-fold cross-validation, training dibagi menjadi k bagian. Setiap bagian sekali menjadi validation fold, model dilatih pada k-1 bagian sisanya, dan skor dirata-ratakan. Ini menilai variasi generalisasi di beberapa pembagian tanpa memakai test set untuk memilih model. Validation fold berbeda dari test akhir yang sudah dipisahkan.

Sel tambahan memakai **hanya X_train/y_train dari holdout sebelumnya**. KFold dengan shuffle cocok sebagai demonstrasi pengamatan yang dapat diperlakukan independen; untuk prediksi masa depan gunakan split waktu, dan untuk pengamatan berkelompok gunakan split kelompok. Scaling, imputasi, seleksi fitur, dan encoding berbasis data perlu di-fit hanya pada bagian training tiap fold, misalnya melalui pipeline.

Scikit-learn memakai skor negatif untuk metrik error tertentu karena skor lebih tinggi berarti lebih baik. Oleh karena itu `neg_root_mean_squared_error` dibalik tanda sebelum dilaporkan sebagai RMSE. Mean/std antar-fold merangkum variasi skor, bukan confidence interval formal dan bukan jaminan performa deployment. Seleksi hiperparameter sebaiknya dinilai dengan desain nested CV atau test akhir yang tidak disentuh jika ingin estimasi yang tidak optimistis.

```python
from sklearn.model_selection import KFold, cross_val_score

regression_folds = KFold(n_splits=5, shuffle=True, random_state=42)
fold_rmse = -cross_val_score(LinearRegression(), X_train, y_train,
                             cv=regression_folds, scoring='neg_root_mean_squared_error')
display(pd.Series(fold_rmse, index=[f'fold_{i}' for i in range(1, 6)], name='validation_RMSE'))
print('Mean RMSE:', fold_rmse.mean())
print('SD RMSE antar-fold:', fold_rmse.std(ddof=1))
assert len(fold_rmse) == 5 and np.isfinite(fold_rmse).all()
print('Interpretasi: skor dihitung di validation fold dari training, bukan di test akhir.')
```
